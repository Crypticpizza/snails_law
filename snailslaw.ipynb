```python
from matplotlib.lines import lineStyles
import numpy as np
import matplotlib.pyplot as plt

n1 = 1 #air
n2 = 1.5 #glass

theta_i_deg = 45 #incident angle
theta_i = np.radians(theta_i_deg)

#snails law gives
sin_theta_r = (n1/n2)*np.sin(theta_i)

if sin_theta_r > 1.0:
  print("total internal reflection")
else:
  theta_r = np.arcsin(sin_theta_r)

theta_r_deg = np.degrees(theta_r)
print(theta_r_deg)


#PLOTTING
plt.figure(figsize=(6,6))
plt.axhline(0, color="black", linewidth = 2, label= "boundary air/glass")

#defining the rays
#incident ray
ray_in_x = np.array([-np.sin(theta_i), 0])
ray_in_y = np.array([np.cos(theta_i), 0])
#refracted ray
ray_ref_x = np.array([0, np.sin(theta_r)])
ray_ref_y = np.array([0, -np.cos(theta_r)])


#plotting the rays
plt.plot(ray_in_x, ray_in_y, color="red", label=f'Incident Ray ({theta_i_deg}°)')
plt.plot(ray_ref_x, ray_ref_y, color="blue", label= "refracted ray")

#formatting the plot
plt.xlim(-1.2, 1.2)
plt.ylim(-1.2, 1.2)
plt.axvline(0, color="grey", linestyle='--', alpha = 0.5)
plt.grid(True, alpha=0.3)
plt.legend()

```
