```python
from matplotlib.lines import lineStyles
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

def update_snails(theta_1_deg):
  n2 = 1 #air
  n1 = 1.33 #water
  n3 = 1.5 #glass

  theta_1 = np.radians(theta_1_deg)

  #snails law gives
  sin_theta_2 = (n1/n2)*np.sin(theta_1)

  if sin_theta_2 > 1.0:
    print("total internal reflection")
    # If total internal reflection occurs, we can't calculate theta_2, so we skip plotting the refracted ray
    plt.figure(figsize=(6,6))
    plt.axhline(0, color="black", linewidth = 2, label= "boundary air/glass")

    #incident ray
    ray_in_x = np.array([-np.sin(theta_1), 0])
    ray_in_y = np.array([np.cos(theta_1), 0])
    plt.plot(ray_in_x, ray_in_y, color="red", label=f'Incident Ray ({theta_1_deg}°)')
    plt.plot(-1*ray_in_x, ray_in_y, color="blue", label=f'Total Internal Reflection')
    plt.xlim(-1.2, 1.2)
    plt.ylim(-1.2, 1.2)
    plt.axvline(0, color="grey", linestyle='--', alpha = 0.5)
    plt.grid(True, alpha=0.3)
    plt.legend()

    return
  else:
    theta_2 = np.arcsin(sin_theta_2)

  theta_2_deg = np.degrees(theta_2)


  #PLOTTING
  plt.figure(figsize=(6,6))
  plt.axhline(0, color="black", linewidth = 2, label= "boundary air/glass")

  #defining the rays
  #incident ray
  ray_in_x = np.array([-np.sin(theta_1), 0])
  ray_in_y = np.array([np.cos(theta_1), 0])
  #refracted ray
  ray_ref_x = np.array([0, np.sin(theta_2)])
  ray_ref_y = np.array([0, -np.cos(theta_2)])


  #plotting the rays
  plt.plot(ray_in_x, ray_in_y, color="red", label=f'Incident Ray ({theta_1_deg}°)')
  plt.plot(ray_ref_x, ray_ref_y, color="blue", label= "refracted ray")

  #formatting the plot
  plt.xlim(-1.2, 1.2)
  plt.ylim(-1.2, 1.2)
  plt.axvline(0, color="grey", linestyle='--', alpha = 0.5)
  plt.grid(True, alpha=0.3)
  plt.legend()

widgets.interact(update_snails, theta_1_deg = widgets.FloatSlider(value= 45.0, min = 0.0, max = 90, step = 0.1))
```
